## 0. 준비

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
np.random.seed(42)
tf.random.set_seed(42)
print(tf.__version__)
```

## 2. 데이터 불러오기 (쓰는 것만 주석 해제)

```python
# --- MNIST / Fashion MNIST (Dense용) ---
(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.mnist.load_data()   # fashion_mnist 로 바꿔도 됨
X_train_full = X_train_full / 255.0
X_test = X_test / 255.0
X_train, X_valid = X_train_full[:-5000], X_train_full[-5000:]
y_train, y_valid = y_train_full[:-5000], y_train_full[-5000:]
print(X_train.shape, X_valid.shape, X_test.shape)

# --- CNN이면 채널 차원 추가 ---
# X_train, X_valid, X_test = X_train[..., np.newaxis], X_valid[..., np.newaxis], X_test[..., np.newaxis]
```

## 3. 모델 (문제 사양대로 수정)

```python
model = keras.Sequential([
    keras.layers.Input(shape=(28, 28)),            # CNN이면 (28, 28, 1)
    keras.layers.Flatten(),
    keras.layers.Dense(300, activation="relu", kernel_initializer="he_normal"),
    # keras.layers.BatchNormalization(),
    keras.layers.Dense(100, activation="relu", kernel_initializer="he_normal"),
    # keras.layers.Dropout(0.2),
    keras.layers.Dense(10, activation="softmax")
])
# CNN 예시:
# keras.layers.Conv2D(32, kernel_size=(3, 3), padding="same", activation="relu"),
# keras.layers.MaxPooling2D(pool_size=(2, 2)),
model.summary()
```

## 4. 컴파일 · 학습

```python
model.compile(loss="sparse_categorical_crossentropy",       # 원-핫이면 categorical_crossentropy
              optimizer=keras.optimizers.SGD(learning_rate=0.01, momentum=0.9, nesterov=True),   # 문제대로
              metrics=["accuracy"])
history = model.fit(X_train, y_train, epochs=10,
                    validation_data=(X_valid, y_valid), verbose=2)
```

## 5. 평가 · 학습 곡선 · 예측

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"테스트 정확도: {test_acc:.4f}")

pd.DataFrame(history.history).plot(figsize=(8, 5))
plt.grid(True); plt.xlabel("epoch"); plt.show()

y_pred = np.argmax(model.predict(X_test[:5]), axis=-1)
print("예측:", y_pred, " 정답:", y_test[:5])
```
